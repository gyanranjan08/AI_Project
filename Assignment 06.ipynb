**Q1. Create a semantic network with visualization using Python**

```python
import networkx as nx
import matplotlib.pyplot as plt

# Create a directed graph
G = nx.DiGraph()

# Add knowledge as relationships
relationships = [
    ("Dog", "is_a", "Animal"),
    ("Cat", "is_a", "Animal"),
    ("Animal", "needs", "Food"),
    ("Dog", "eats", "Food"),
    ("Cat", "eats", "Food"),
    ("Dog", "has", "Tail"),
    ("Cat", "has", "Tail"),
    ("Dog", "can", "Bark"),
    ("Cat", "can", "Meow")
]

# Add nodes and edges
for subject, relation, object_ in relationships:
    G.add_edge(subject, object_, relation=relation)

# Draw the semantic network
pos = nx.spring_layout(G, seed=42)

nx.draw(
    G,
    pos,
    with_labels=True,
    node_color="lightblue",
    node_size=2500,
    font_size=10,
    arrows=True
)

# Display relationship labels
edge_labels = nx.get_edge_attributes(G, "relation")
nx.draw_networkx_edge_labels(
    G,
    pos,
    edge_labels=edge_labels,
    font_size=9
)

plt.title("Semantic Network")
plt.show()
```
